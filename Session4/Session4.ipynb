```python
# linear function
def poly_1(X):
    y=4*X+5
    return y
def poly_2(X):
    y=6*(X**2)+4*X+5
    return y
def poly_3(X):
    y=8*(X**3)+6*(X**2)+4*X+5
    return y
```
