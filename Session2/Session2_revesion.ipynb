```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
df=pd.read_csv("Advertising.csv")
```

```python
df
```

```python
df.drop('Unnamed: 0', axis=1, inplace=True)
```

```python
# checking linear relationship between features and target variable
```

```python
for feature in df.columns[:-1]:
    plt.figure()
    plt.scatter(df[feature], df['Sales'])
    plt.title(f'Scatter plot of {feature} vs Sales')
    plt.xlabel(feature)
    plt.ylabel('Sales')
    plt.show()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True)
```

```python
# separate input and output features
```

```python
x=df[['TV']]  #independent variable
y=df['Sales'] # dependent variable
```

```python
# perform train test split 
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
#  create a linear regression model
```

```python
lr= LinearRegression()
# this object is going to call ols method to find the best fit line for our data
```

```python
lr.fit(x_train, y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
# now we can use the model to make predictions on the trained data
```

```python
y_pred=lr.predict(x_train)
```

```python
plt.scatter(x_train, y_train)
plt.plot(x_train, y_pred, color='red')
plt.show()
```

```python
# now perform the prediction on the test data
```

```python
Y_pred=lr.predict(x_test)
```

```python
mse=mean_squared_error(y_test, Y_pred)
emse=root_mean_squared_error(y_test, Y_pred)
mae=mean_absolute_error(y_test, Y_pred)
r2=r2_score(y_test, Y_pred)
```

```python
print(mse)
print(emse)
print(mae)
print(r2)
```

```python
x=df[['Radio']]  #independent variable
y=df['Sales'] # dependent variable

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)

lr= LinearRegression()
lr.fit(x_train, y_train)
lr.coef_
lr.intercept_

y_pred=lr.predict(x_train)

plt.scatter(x_train, y_train)
plt.plot(x_train, y_pred, color='red')
plt.show()

Y_pred=lr.predict(x_test)

mse=mean_squared_error(y_test, Y_pred)
emse=root_mean_squared_error(y_test, Y_pred)
mae=mean_absolute_error(y_test, Y_pred)

print(mse)
print(emse)
print(mae)
```
